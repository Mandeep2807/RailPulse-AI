```python

from pathlib import Path
import pandas as pd

base = Path("../")
tables_dir = base / "outputs" / "tables"
figures_dir = base / "outputs" / "figures"

important_tables = [
    "chronological_model_comparison.csv",
    "unseen_train_model_comparison.csv",
    "feature_ablation_results.csv",
    "final_reserved_group_model_comparison.csv",
    "final_reserved_group_category_analysis.csv",
    "permutation_feature_importance.csv",
    "sensitivity_experiment_results.csv",
    "delay_categories.csv",
    "daily_delay_summary.csv",
    "cluster_selection_metrics.csv",
    "cluster_profiles.csv",
]

print("=" * 65)
print("RAILPULSE AI — FINAL RESULTS INVENTORY")
print("=" * 65)

for filename in important_tables:
    path = tables_dir / filename

    if path.exists():
        df = pd.read_csv(path)
        print(f"\n[FOUND] {filename}")
        print("Shape:", df.shape)
        print("Columns:", df.columns.tolist())
    else:
        print(f"\n[NOT FOUND] {filename}")

print("\n" + "=" * 65)
print("AVAILABLE FIGURES")
print("=" * 65)

if figures_dir.exists():
    figures = sorted(figures_dir.glob("*.png"))
    for fig in figures:
        print(fig.name)

    print("\nTotal PNG figures:", len(figures))
else:
    print("Figures directory not found.")
```

```python

from pathlib import Path
import pandas as pd

base = Path("../")
tables = base / "outputs" / "tables"

# Load the key experimental results
files = {
    "Chronological Models": "chronological_model_comparison.csv",
    "Unseen Train Models": "unseen_train_model_comparison.csv",
    "Feature Ablation": "feature_ablation_results.csv",
    "Final Reserved Evaluation": "final_reserved_group_model_comparison.csv",
    "Delay Category Analysis": "final_reserved_group_category_analysis.csv",
    "Feature Importance": "permutation_feature_importance.csv",
    "Sensitivity Experiment": "sensitivity_experiment_results.csv",
    "Delay Categories": "delay_categories.csv",
    "PCA Variance": "pca_explained_variance.csv",
    "KMeans Metrics": "kmeans_cluster_selection_metrics.csv",
    "KMeans Profiles": "kmeans_cluster_profiles.csv",
    "Delay-Focused Metrics": "delay_focused_cluster_metrics.csv",
    "Delay-Focused Profiles": "delay_focused_cluster_profiles.csv",
}

results = {}

for title, filename in files.items():
    path = tables / filename
    if not path.exists():
        print("MISSING:", filename)
        continue

    results[title] = pd.read_csv(path)
    print(f"[OK] {title}: {results[title].shape}")

# Save a compact inventory of all consolidated tables
inventory = pd.DataFrame([
    {
        "Section": title,
        "Source_File": files[title],
        "Rows": len(df),
        "Columns": len(df.columns)
    }
    for title, df in results.items()
])

inventory.to_csv(
    tables / "railpulse_final_results_inventory.csv",
    index=False
)

print("\n" + "=" * 65)
print("FINAL MODEL PERFORMANCE")
print("=" * 65)
print(results["Final Reserved Evaluation"].to_string(index=False))

print("\n" + "=" * 65)
print("SENSITIVITY ANALYSIS")
print("=" * 65)
print(results["Sensitivity Experiment"].to_string(index=False))

print("\n" + "=" * 65)
print("PCA RESULTS")
print("=" * 65)
print(results["PCA Variance"].to_string(index=False))

print("\n" + "=" * 65)
print("DELAY-FOCUSED CLUSTER PROFILES")
print("=" * 65)
print(results["Delay-Focused Profiles"].to_string(index=False))

print("\n" + "=" * 65)
print("CONSOLIDATED INVENTORY SAVED")
print(inventory.to_string(index=False))
```
